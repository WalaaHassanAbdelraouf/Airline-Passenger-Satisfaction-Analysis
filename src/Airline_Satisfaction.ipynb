## Data Cleaning

```python
# Checking for missing values

data.isnull().sum()
```

```python
# dropping the unnamed column as it's an index
df = data.drop(['Unnamed: 0'], axis = 1)
df
```

```python
# Handling the missing values in 'Arrival Delay in Minutes' feature

df['Arrival Delay in Minutes']= df['Arrival Delay in Minutes'].fillna(df['Arrival Delay in Minutes'].median())
```

```python
df.isnull().sum()
```

```python
class_count = df['Class'].value_counts()
```

```python
# Displaying the number of passengers in each class

plt.pie(class_count, data = df, labels = class_count.index,
        shadow =True, explode =(0.05,0.05,0.05), colors=['#add8e6', '#6495ed', '#0000ff'], autopct='%1.1f%%', startangle=90)
plt.title("No of Passengers per Class Types")
plt.show()
```

- The number of travellers in business class is the highest, and this indicates a significant portion of passengers prefer the premium experience offered by business class.
- The economy class attracts a considerable number of passengers, while the Eco plus passengers seem to be less popular which indicates that passengers either choose the premium experience of business class or stick to the more affordable economy one.

```python
travel_type = df['Type of Travel'].value_counts()
```

```python
# Displaying the proportion of each type of travelling

plt.pie(travel_type, labels = travel_type.index, shadow =True, 
        explode =(0.1, 0), colors=['#add8e6', '#ffb6c1'], autopct='%1.1f%%', startangle=90 )
plt.title("Business Vs Personal Travel")
plt.show()
```

- The number of travellers on business trips is greater than those travelling on personal trips, and this indicates that the airline is likely catering more to business travelers, which could influence various aspects of services and operations.

```python
# Displaying the proportion of loyal and disloyal customers of the whole passengers

customer_type = df['Customer Type'].value_counts()
```

```python
# Plotting the donut chart 
plt.pie(customer_type, labels=customer_type.index, autopct='%1.1f%%',colors = ['#add8e6', '#ffb6c1'],
        shadow =True, startangle=90, wedgeprops={'width':0.5}, explode=(0.1,0))

plt.title('Loyal Vs Disloyal Customers')
plt.show()
```

The loyal customers have a high proportion compared to disloyal customers. and this may indicates that this Airline has satisfactory experience and services.

```python
# Displaying the Satisfaction level of passengers

satisfied_counts = df['satisfaction'].value_counts()
```

```python
plt.pie(satisfied_counts, labels = satisfied_counts.index, shadow =True, 
        explode =(0.1, 0), colors=['#add8e6', '#ffb6c1'], autopct='%1.1f%%', startangle=130 )
plt.title("Satisfaction Level")
plt.show()
```

The number of satisfied passengers tend to be a little bit lower than the dissatisfied ones, and this indicates that the dataset is imbalanced.

```python
# Satisfaction by Class 
plt.figure(figsize=(12, 6))
sns.countplot(x='Class', hue='satisfaction', data=df, palette='husl')
plt.title('Satisfaction Level by Class')
plt.xlabel('Class')
plt.ylabel('Count')
plt.show()
```

**Class Type:**
- The satisfaction level varies significantly between the different classes ( Economy, Eco-plus, and Business).
- The number of satisfied passengers in the business class has a higher range than from Economy and Eco-Plus classes, and this indicates that the quality of service and experience may differ based on the class of travelling.


```python
# Satisfaction by Customer Type
plt.figure(figsize=(10, 6))
sns.countplot(x='Customer Type', hue='satisfaction', data=df, palette=['#add8e6', '#ffb6c1'])
plt.title('Satisfaction by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Count')
plt.show()


```

**Customer Type:**
- From above, the loyal customers shows higher satisfaction compared to disloyal customers, and it suggests that frequent passengers tend to have better experiences or are more satisfied with the services, but in general the number of dissatisfied passengers are greater than the satisfied ones in both cases.

```python
# Displaying the satisfaction level by gender

plt.figure(figsize=(10, 6))
sns.countplot(x='Gender', hue='satisfaction', data=df, palette='husl')
plt.title('Gender Distribution by Satisfaction')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

**Gender Satisfaction Level:**
 - The satisfaction levels differ by gender, and this may indicate that different genders have different experiences. 

**Recommendation:**
- Tailoring services to better meet the needs of different gender groups.


```python
# Displaying the distribution of Age 

plt.figure(figsize=(8,6))
sns.histplot(data = df, x='Age', bins=50, kde=True)
plt.title('Age Distribution')
plt.show()
```

**Most of passengers' ages ranges between 21 and 60**


```python
# Displaying the satisfaction level by age groups

# Binning
bins = [0, 20, 40, 60, 80, 100]
labels = ['0-20', '21-40', '41-60', '61-80', '81+']
df['Age Group'] = pd.cut(df['Age'], bins=bins, labels=labels)

# Plotting
plt.figure(figsize=(12, 6))
sns.countplot(x='Age Group', hue='satisfaction', data=df, palette = ['#add8e6', '#6495ed'])
plt.title('Satisfaction by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.show()
```

**Age Groups Satisfaction:**
- These age groups have varying levels of satisfaction. 
- The younger passengers tend to be less satisfied compared to older ones, which can guide targeted improvements for specific age groups.

```python
features = ['Inflight wifi service', 'Ease of Online booking', 'Inflight entertainment', 
           'On-board service', 'Baggage handling', 'Checkin service', 'Cleanliness']

# Defining the number of rows and columns
n_ratings = len(features)
ncols = 3
nrows = 3

fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=(14, 8))

# Flatten axes 
if nrows == 1 or ncols == 1:
    axes = [axes]
else:
    axes = axes.flatten()

# Plot each rating on a subplot
for i in range(n_ratings):
    sns.boxplot(x='satisfaction', y=features[i], data=df, ax=axes[i])
    axes[i].set_title(f'{features[i]} vs Satisfaction')
    axes[i].set_xlabel('Satisfaction')
    axes[i].set_ylabel(features[i])



plt.tight_layout()
plt.show()
```

**1. Inflight Wi-Fi Service vs. Satisfaction:**
- The boxplot shows a higher median and tighter interquartile range for the 'satisfied' category compared to the "dissatisfied" one, and it suggests that passengers who rated the inflight Wi-Fi service higher are generally more satisfied.

**2. Ease of Online Booking vs. Satisfaction:**
- The boxplot shows a higher median for the "satisfied" group and it indicates that the ease of online booking is positively correlated with passenger satisfaction. 

**3. Inflight Entertainment vs. Satisfaction:**
- The boxplot shows a clear distinction between the "satisfied" and "dissatisfied" groups, with lower ratings for inflight entertainment correlating with lower satisfaction, it indicates that inflight entertainment plays a significant role in passenger satisfaction.

**4. On-board Service vs. Satisfaction:**
- The higher median and more concentrated IQR for the "satisfied" group would suggest that good on-board service is strongly associated with satisfaction. 

**5. Baggage Handling vs. Satisfaction:**
- The boxplot shows a higher median in the "satisfied" category and it indicates that the efficient baggage handling is a key factor in customer satisfaction.


**6. Check-in Service vs. Satisfaction:**
- A higher median for the "satisfied" group may indicate that smoother check-in processes are linked to higher satisfaction.


**7.Cleanliness vs. Satisfaction:**
- Cleanliness is often a critical factor in customer experience. A noticeable difference between the "satisfied" and "dissatisfied" groups, with higher median for the satisfied group, would indicate that cleanliness is a key determinant of overall satisfaction.

**Most of the plots show similar trends, and this indicates that these service aspects are consistently important across different passengers in determining overall satisfaction.**


# Suggestions:


1. The satisfaction level is notably lower in Economy class compared to Business Class,So we should consider improvements in service quality, comfort, or amenities in Economy class.

2. The different genders and age groups show different satisfaction levels, then we should tailor marketing and service offerings to meet the specific needs and preferences of these groups.

3. Services like inflight entertainment, ease of online booking and cleanliness that show a strong impact on satisfaction should be regularly monitored and improved.
